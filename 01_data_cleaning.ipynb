```python
import pandas as pd

df = pd.read_csv('data/raw/train_FD001.txt', sep = r'\s+', header = None)
df.head()
```

```python
# Give proper column names instead of 0,1,2,3...
columns = ['unit_number', 'time_in_cycles', 'op_setting_1', 'op_setting_2', 'op_setting_3'] + [f'sensor_{i}' for i in range(1, 22)]
df.columns = columns

df.head()
```

```python
print(df.shape)
df.info()
```

```python
print(df.isnull().sum().sum())  # total missing values across whole dataframe
print(df.describe())  # statistical summary - look for sensors with 0 variation
```

```python
# Find columns where standard deviation is 0 (no variation at all)
constant_cols = df.columns[df.std() == 0]
print("Constant (useless) columns:", list(constant_cols))
```

```python
# Drop constant columns - they add no predictive value
df_clean = df.drop(columns=constant_cols)
print("Shape before:", df.shape)
print("Shape after:", df_clean.shape)
df_clean.head()
```

```python
# Rolling average of sensor_2 (temperature-like sensor) over last 5 cycles, per engine
df_clean = df_clean.sort_values(['unit_number', 'time_in_cycles'])
df_clean['sensor_2_roll_mean_5'] = (
    df_clean.groupby('unit_number')['sensor_2']
    .transform(lambda x: x.rolling(window=5, min_periods=1).mean())
)

df_clean[['unit_number', 'time_in_cycles', 'sensor_2', 'sensor_2_roll_mean_5']].head(10)
```

```python
# Calculate RUL: how many cycles remain until failure, per engine
max_cycles = df_clean.groupby('unit_number')['time_in_cycles'].transform('max')
df_clean['RUL'] = max_cycles - df_clean['time_in_cycles']

df_clean[['unit_number', 'time_in_cycles', 'RUL']].head(10)
```

```python
# Rolling standard deviation - captures increasing instability before failure
df_clean['sensor_2_roll_std_5'] = (
    df_clean.groupby('unit_number')['sensor_2']
    .transform(lambda x: x.rolling(window=5, min_periods=1).std())
)

# Fill the first-row NaN (std of 1 value is undefined) with 0
df_clean['sensor_2_roll_std_5'] = df_clean['sensor_2_roll_std_5'].fillna(0)

df_clean.shape
```

```python
cols_to_drop = ['op_setting_3', 'sensor_1', 'sensor_10', 'sensor_18', 'sensor_19']
cols_present = [c for c in cols_to_drop if c in df_clean.columns]
if cols_present:
    df_clean = df_clean.drop(columns=cols_present)
    print(f"Dropped: {cols_present}")
else:
    print("No constant columns present — already clean.")

print("Final shape:", df_clean.shape)
```

```python
df_clean.to_csv('data/processed/clean_data.csv', index=False)
print("Saved successfully!")
```
